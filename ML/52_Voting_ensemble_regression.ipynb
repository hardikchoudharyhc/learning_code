Import the libraries.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

Load the dataset.

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)

df = housing.frame

df_sample = df.sample(n=1000, random_state=42)

X = df_sample.drop('MedHouseVal', axis=1)
y = df_sample['MedHouseVal']

print(X.shape, y.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

Find the scores for LR, DT and SVR algorithms.

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
     

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

Voting ensemble on above estimators.

```python
from sklearn.ensemble import VotingRegressor
     

vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

Find the best score for the weights.

```python
best_score = 0
best_weights = None

for i in range(1, 4):
    for j in range(1, 4):
        for k in range(1, 4):

            vr = VotingRegressor(
                estimators=estimators,
                weights=[i, j, k]
            )

            scores = cross_val_score(
                vr, X, y,
                cv=10,
                scoring='r2'
            )

            mean_score = scores.mean()

            if mean_score > best_score:
                best_score = mean_score
                best_weights = [i, j, k]

            print(
                f"for i={i}, j={j}, k={k}",
                round(mean_score, 6)
            )

print("\nBest score:", round(best_score, 6))
print("Best weights:", best_weights)
```

Find the score for DT with diff max_depth.

```python
# using the same algorithm

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
     
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
     

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

Score for voting ensemble.

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
