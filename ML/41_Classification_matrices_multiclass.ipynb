```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Datasets/Iris.csv')
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Species'] = encoder.fit_transform(df['Species'])
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.drop('Species', axis=1), df['Species'], test_size=0.2, random_state=1)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()

clf1.fit(x_train, y_train)
clf2.fit(x_train, y_train)

y_pred1 = clf1.predict(x_test)
y_pred2 = clf2.predict(x_test)
import numpy as np

print(np.array_equal(y_pred1, y_pred2))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred1))
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred2))
```

```python
print("Logistic Regression Confusion Matrix:\n", pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,3))))

print("Decision Tree Confusion Matrix\n",pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,3))))
```

```python
result = pd.DataFrame()
result['Actual Label'] = y_test
result['Logistic Regression Prediction'] = y_pred1
result['Decision Tree Prediction'] = y_pred2
result.sample(10)
```

```python
from sklearn.metrics import precision_score,recall_score
precision_score(y_test,y_pred1,average=None)
```

```python
recall_score(y_test,y_pred1,average=None)
```

```python
df2 = pd.read_csv('Datasets/fashion-mnist.csv')
df2.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df2.iloc[:,1:],df2.iloc[:,0],test_size=0.2,random_state=2)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

clf1 = LogisticRegression()
clf1.fit(X_train,y_train)
y_pred1 = clf1.predict(X_test)

clf2 = DecisionTreeClassifier()
clf2.fit(X_train,y_train)
y_pred2 = clf2.predict(X_test)

from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,10)))
```

```python
print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,10)))
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
print("Precision of Logistic Regression\n",precision_score(y_test,y_pred1,average="weighted"))
print("Recall of Logistic Regression\n",recall_score(y_test,y_pred1,average="weighted"))
print("F1 Score of Logistic Regression\n",f1_score(y_test,y_pred1,average="weighted"))
```

```python

from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred1))
```
