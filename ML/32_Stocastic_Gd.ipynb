Import the libraries.

```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import time
```

Load the premade dataset from sklearn.

```python
X,y = load_diabetes(return_X_y=True)
print(X.shape, y.shape)
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

Train the model and find r2 score.

```python
model = LinearRegression()
model.fit(X_train, y_train)
print(model.coef_, model.intercept_)
```

```python
y_pred = model.predict(X_test)
r2_score(y_test, y_pred)
```

Self made Stocastic GD class.

```python
class SelfMadeGD:
    def __init__(self, learning_rate, epochs):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):
        self.coef_ = np.zeros(X_train.shape[1])
        self.intercept_ = 0

        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                index = np.random.randint(0, X_train.shape[0])

                y_hat = np.dot(X_train[index], self.coef_) + self.intercept_

                error = y_train[index] - y_hat

                intercept_derivative = -2 * error
                self.intercept_ -= self.lr * intercept_derivative

                coef_derivative = -2 * X_train[index] * error
                self.coef_ -= self.lr * coef_derivative

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
    
```

Train the model and find the r2 score.

```python
sgd = SelfMadeGD(learning_rate=0.01, epochs=100)
sgd.fit(X_train, y_train)
print(sgd.coef_, sgd.intercept_)
```

```python
print(r2_score(y_test, sgd.predict(X_test)))
```

Train the model with SDGRegressor function of sklearn and compare the results.

```python
from sklearn.linear_model import SGDRegressor
sgd_regressor = SGDRegressor(learning_rate='constant', eta0=0.03, max_iter=100)
sgd_regressor.fit(X_train, y_train)
print(sgd_regressor.coef_, sgd_regressor.intercept_)
```

```python
print(r2_score(y_test, sgd_regressor.predict(X_test)))
```
