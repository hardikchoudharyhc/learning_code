Working with dates.

Convert object type date into datetime64.

```python
date['date'] = pd.to_datetime(date['date'])
date.info()
```

Extract year.

```python
date['date_year'] = date['date'].dt.year
date.sample(5)
```

Extract semester and quater.

```python
date['quarter'] = date['date'].dt.quarter
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

Extract month number and name.

```python
date['date_month_no'] = date['date'].dt.month
date.sample(5)
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.sample(5)
```

Extract week of the year.

```python
# date['date_week'] = date['date'].dt.week
date['date_week'] = date['date'].dt.isocalendar().week.astype('int32')
```

```python
date['date_day'] = date['date'].dt.day
date.sample(5)
```

```python
date['date_dow'] = date['date'].dt.day_name() # Day of week.
date.sample(5)
```

```python
# day of week - name
date['date_dow_name'] = date['date'].dt.day_name()

# date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0) # Using numpy where function to find if saturday and sunday are in the column.
```

Extract time passed between dates.

```python
import datetime

today = datetime.datetime.today()
# today.year
# today.month
# today.day
# today.weekday() # Monday is 0 and Sunday is 6.
today
```

Find the days passed.

```python
(today - date['date']).dt.days # Difference in days between today and the date column.
```

Find the month passed.

```python
(today.year - date['date'].dt.year) * 12 + (today.month - date['date'].dt.month)
# or np.round((today -date['date']) / np.timedelta64(1, 'M'),0)
```

Work on the time dataset.

```python
time.info()
```

Change datatype to datetime64.

```python
# Converting to datetime datatype
time['date'] = pd.to_datetime(time['date'])
time.info()
```

Create new column for hour, min, sec.

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

Extract time part from date column.

```python
time['time'] = time['date'].dt.time
time.head()
```

Find the time difference.

```python
today - time['date'] # Difference between today and the date column in datetime format.

(today - time['date'])/np.timedelta64(1,'h') # Difference in hours between today and the date column.
(today - time['date'])/np.timedelta64(1,'m') # Difference in minutes between today and the date column.
(today - time['date'])/np.timedelta64(1,'s') # Difference in seconds between today and the date column.
```
