Import the library.

```python
import numpy as np
import pandas as pd
```

Read the dataset.

```python
df = pd.read_csv("Datasets/titanic.csv")
df.sample(5)
```

```python
df['number'].unique()
```

Bar plot for count of number column.

```python
fig = df['number'].value_counts().plot(kind='bar')
fig.set_xlabel("Number of people")
fig.set_ylabel("Count")
```

Extract the numerical and categorical part from number column.

```python
# extract numerical part
df['number_numerical'] = pd.to_numeric(df["number"],errors='coerce',downcast='integer')
# extract categorical part
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'],np.nan)
df.head()
```

Find the unique entries in Ticket column.

```python
df['Ticket'].unique()
```

Find the unique entries in Cabin column.

```python
df['Cabin'].unique()
```

Extract the numerical and categorical part from Cabin column.

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)') # captures numerical part
df['cabin_cat'] = df['Cabin'].str[0] # captures the first letter

df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

Extract the numerical and categorical part from Ticket column.

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],errors='coerce',downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,df['ticket_cat'])
df.head()
```
