Import the libraries.

```python
import numpy as np

from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

Load the premade database from sklearn and split the data into train and test. 

```python
X,y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=13)
```

Train the model with linear regression.

```python
model = LinearRegression()
model.fit(X_train, y_train)
model.coef_, model.intercept_
```

Find the r2 score.

```python
y_pred = model.predict(X_test)
r2_score(y_test, y_pred)
```

Make a self made GD function with variable m.

```python
class SelfMadeGD:
    def __init__(self, learning_rate, epochs):
        self.m = None
        self.c = None
        self.lr = learning_rate
        self.epochs = epochs
    
    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.zeros(X_train.shape[1])

        for i in range(self.epochs):

            y_hat = self.intercept_ + np.dot(X_train, self.coef_) # Vectorization 
            intercept_deivative = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_ - self.lr * intercept_deivative

            coef_derivative = -2 * np.dot(X_train.T, (y_train - y_hat)) / X_train.shape[0]
            self.coef_ = self.coef_ - self.lr * coef_derivative
        
    def predict(self, X_test):
        return self.intercept_ + np.dot(X_test, self.coef_)
```

Train the model and find the r2 score.

```python
gdr = SelfMadeGD(learning_rate=0.01, epochs=100)
gdr.fit(X_train, y_train)
print(gdr.coef_, gdr.intercept_)
```

```python
y_pred = gdr.predict(X_test)
r2_score(y_test, y_pred)
```
