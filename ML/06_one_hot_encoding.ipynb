Importing the libraries.

```python
import numpy as np
import pandas as pd
```

Reading the dataset.

```python
df = pd.read_csv('Datasets/cars.csv')
print(df.sample(5))
```

Finding the all different values in a column.

```python
print(df['brand'].nunique())
print(df['brand'].value_counts())
```

```python
print(df['fuel'].unique())
print(df['fuel'].value_counts())
```

```python
print(df['owner'].unique())
print(df['owner'].value_counts())
```

.get_dummies() - It converts categorical values into separate indicator (0/1) columns for One - Hot encoding



```python
pd.get_dummies(df, columns=['fuel', 'owner'])
```

```python
pd.get_dummies(df, columns=['fuel', 'owner'], drop_first=True)
```

Spliting the data into training and testing part.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:,0:4], df.iloc[:,-1], test_size=0.2, random_state=42)

x_train.head()
```

Importing OneHotencoder class fomr sklearn.preprocessing ad fit_transforming train data (test data should be avoided).

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder()

x_train_new = ohe.fit_transform(x_train[['fuel', 'owner']]).toarray()
x_test_new = ohe.transform(x_test[['fuel', 'owner']]).toarray()

x_train_new[:5]
x_test_new[:5]  
# x_train_new.shape
```

Concatenting the array into one single numpy array.

```python
np.hstack((x_train[['brand','km_driven']].values,x_train_new))
```

Removing the brand with value less than a threshold and merging them into a single category as 'Other'

```python
count = df['brand'].value_counts()
print(df['brand'].nunique())
threshold = 100
```

```python
replace_brand = count[count<=threshold].index
pd.get_dummies(df['brand'].replace(replace_brand, 'Other')).sample(5)
```
