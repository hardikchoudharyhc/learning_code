Import the libraries.

```python
import numpy as np
import pandas as pd
```

Create a dataframe.

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
df
```

Self made functions.

Function to initialize the parameters.

```python
def init_params(layer_dims): # Function to initialize parameters
    np.random.seed(3)
    parameters = {}
    L = len(layer_dims) # number of layers in the network
    
    for l in range(1, L):
        parameters['W' + str(l)] = np.random.randn(layer_dims[l], layer_dims[l-1]) * 0.01
        parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))
    return parameters
```

```python
init_params([2, 2, 1])
```

Z = W.A + b

```python
def linear_forward(A_prev,W,b):
    Z = np.dot(W,A_prev) + b
    return Z
```

Function for forward propogation.

```python
def forward_propagation(X, parameters):
    A = X
    L = len(parameters) // 2 # number of layers in the neural network
    
    for l in range(1, L + 1):
        A_prev = A
        Wl = parameters['W' + str(l)]
        bl = parameters['b' + str(l)]
        print("A"+str(l-1)+": ", A_prev)
        print("W"+str(l)+": ", Wl)
        print("b"+str(l)+": ", bl)
        print("--"*20)

        A = linear_forward(A_prev, Wl, bl)
        print("A"+str(l)+": ", A)
        print("**"*20)
    return A,A_prev
```

Find the y_hat for a point.

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = init_params([2,2,1])

y_hat,A1 = forward_propagation(X, parameters)
```

Update the paramters using GD.

```python
def update_parameters(parameters,y,y_hat,A1,X):
  error = (y - y_hat).item()

  parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * error * A1[0][0])
  parameters['W2'][0][1] = parameters['W2'][0][1] + (0.001 * 2 * error * A1[1][0])
  parameters['b2'][0][0] = parameters['b2'][0][0] + (0.001 * 2 * error)

  parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * error * parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * error * parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * error * parameters['W2'][0][0])

  parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * error * parameters['W2'][0][1]*X[0][0])
  parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * error * parameters['W2'][0][1]*X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * error * parameters['W2'][0][1])

  return parameters
```

```python
update_parameters(parameters,y,y_hat,A1,X)
```

Forward propogation on points of x and find y_pred than update parameters.

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = init_params([2,2,1])

y_hat,A1 = forward_propagation(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

Repeat the process for epochs.

```python
# epochs implementation

parameters = init_params([2,2,1])
epochs = 50

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):

    X = df[['cgpa', 'profile_score']].values[j].reshape(2,1) # Shape(no of features, no. of training example)
    y = df[['lpa']].values[j][0]

    # Parameter initialization


    y_hat,A1 = forward_propagation(X,parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters,y,y_hat,A1,X)

    Loss.append((y-y_hat)**2)

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())

parameters
```

Demonstration using tenserflow's function.

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

Making layers structure.

```python
model = Sequential()

model.add(Dense(2, input_dim=2, activation='linear'))
model.add(Dense(1, activation='linear'))    
model.summary()
```

Get the default weights of the model.

```python
model.get_weights()
```

Set the weights to same as self initialized one.

```python
new_weights = [np.array([[0.01, 0.01], [0.01, 0.01]], dtype=np.float32), 
               np.array([0., 0.], dtype=np.float32), 
               np.array([[0.01], [0.01]], dtype=np.float32), 
               np.array([0.], dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

Set the lr of optimizer as same as self made and compile the model.

```python
optimizer = keras.optimizers.SGD(learning_rate=0.001)
model.compile(loss='mean_squared_error', optimizer=optimizer)
```

Train the model and print the loss.

```python
model.fit(df[['cgpa', 'profile_score']].values, df[['lpa']].values, epochs=75, verbose=1,batch_size=1)
```
